```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Sales_data.csv")
df
```

```python
clean_df = df.copy()
```

```python
clean_df.isnull().sum()
```

```python
missing_percentage = (clean_df.isnull().sum() / len(clean_df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
numeric_columns = clean_df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if clean_df[column].isnull().sum() > 0:
        clean_df[column] = clean_df[column].fillna(
            clean_df[column].median()
        )
```

```python
categorical_columns = clean_df.select_dtypes(include="object").columns

for column in categorical_columns:
    if clean_df[column].isnull().sum() > 0:
        clean_df[column] = clean_df[column].fillna(
            clean_df[column].mode()[0]
        )
```

```python
clean_df.isnull().sum()
```
