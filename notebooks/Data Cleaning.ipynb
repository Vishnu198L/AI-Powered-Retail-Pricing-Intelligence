```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/sales_data.csv")
df
```

```python
clean_df = df.copy()
```

```python
clean_df.isnull().sum()
```

```python
missing_percentage = (clean_df.isnull().sum() / len(clean_df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
numeric_columns = clean_df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if clean_df[column].isnull().sum() > 0:
        clean_df[column] = clean_df[column].fillna(
            clean_df[column].median()
        )
```

```python
categorical_columns = clean_df.select_dtypes(include=["object", "string", "category"]).columns

for column in categorical_columns:
    if clean_df[column].isnull().sum() > 0:
        clean_df[column] = clean_df[column].fillna(
            clean_df[column].mode()[0]
        )
```

```python
clean_df.isnull().sum()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop_duplicates()
```

```python
clean_df["Date"].dtype
```

```python
clean_df["Date"] = pd.to_datetime(clean_df["Date"],errors="coerce")
```

```python
for column in numeric_columns:
    print(
        column,
        "zero values:",
        (clean_df[column] == 0).sum()
    )
```
