```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
```

```python
df["Month"] = df["Date"].dt.month
```

```python
df["Day"] = df["Date"].dt.day
```

```python
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

```python
df["Quarter"] = df["Date"].dt.quarter
```

```python
df[
    [
        "Date",
        "Year",
        "Month",
        "Day",
        "DayOfWeek",
        "Quarter"
    ]
].head()
```

```python
df["Price_Difference"] = (
    df["Price"] - df["Competitor Pricing"]
)
```

```python
df["Price_Ratio"] = (
    df["Price"] / df["Competitor Pricing"]
)
```

```python
df["Price_Difference_Percent"] = (
    (df["Price"] - df["Competitor Pricing"])
    / df["Competitor Pricing"]
) * 100
```

```python
df["Discount"].describe()
```

```python
df["Effective_Price"] = df["Price"] * (1 - df["Discount"] / 100)
```

```python
df["Stock_Status"] = np.where(
    df["Inventory Level"] == 0,
    "Out_of_Stock",
    "In_Stock"
)
```

```python
df["Inventory_to_Order_Ratio"] = (
    df["Inventory Level"] /
    (df["Units Ordered"] + 1)
)
```

```python
categorical_columns = [
    "Category",
    "Region",
    "Weather Condition",
    "Seasonality",
    "Promotion",
    "Epidemic",
    "Stock_Status"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].unique())
```
