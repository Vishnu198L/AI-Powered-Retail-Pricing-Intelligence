```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("../data/processed/feature_engineered_sales_data.csv")
df
```

```python
pd.set_option("display.max_columns", None)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df = df.sort_values("Date").reset_index(drop=True)
```

```python
df["Date"].is_monotonic_increasing
```

```python
y = df["Demand"]
```

```python
y.head()
```

```python
y.describe()
```

```python
feature_columns = [
    "Category",
    "Region",
    "Inventory Level",
    "Price",
    "Discount",
    "Weather Condition",
    "Promotion",
    "Competitor Pricing",
    "Seasonality",
    "Epidemic",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
     "Price_Difference",
    "Price_Difference_Percent",
    "Stock_Status",
    "Inventory_to_Order_Ratio"
]
```

```python
X = df[feature_columns].copy()
```

```python
X.head()
```

```python
X.shape
```

```python
X.isnull().sum()
```

```python
y.isnull().sum()
```

```python
split_index = int(len(X) * 0.80)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
categorical_features = X_train.select_dtypes(
    include=["object",'string']
).columns.tolist()
categorical_features
```

```python
numerical_features = X_train.select_dtypes(
    include=[np.number]
).columns.tolist()
numerical_features
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)
```

```python
linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)
```

```python
linear_model.fit(X_train, y_train)
```

```python
y_pred_linear = linear_model.predict(X_test)
```

```python
mae_linear = mean_absolute_error(
    y_test,
    y_pred_linear
)

rmse_linear = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_linear
    )
)

r2_linear = r2_score(
    y_test,
    y_pred_linear
)
```

```python
print("Linear Regression")
print("MAE :", mae_linear)
print("RMSE:", rmse_linear)
print("R²  :", r2_linear)
```

```python
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
random_forest_model.fit(
    X_train,
    y_train)
```

```python
y_pred_rf = random_forest_model.predict(
    X_test
)
```

```python
mae_rf = mean_absolute_error(
    y_test,
    y_pred_rf
)

rmse_rf = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_rf
    )
)

r2_rf = r2_score(
    y_test,
    y_pred_rf
)
```

```python
print("Random Forest")
print("MAE :", mae_rf)
print("RMSE:", rmse_rf)
print("R²  :", r2_rf)
```

```python
gradient_boosting_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=200,
                learning_rate=0.05,
                max_depth=3,
                random_state=42
            )
        )
    ]
)
```

```python
gradient_boosting_model.fit(
    X_train,
    y_train
)
```

```python
y_pred_gb = gradient_boosting_model.predict(
    X_test
)
```

```python
mae_gb = mean_absolute_error(
    y_test,
    y_pred_gb
)

rmse_gb = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_gb
    )
)

r2_gb = r2_score(
    y_test,
    y_pred_gb
)
```

```python
print("Gradient Boosting")
print("MAE :", mae_gb)
print("RMSE:", rmse_gb)
print("R²  :", r2_gb)
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        mae_linear,
        mae_rf,
        mae_gb
    ],
    "RMSE": [
        rmse_linear,
        rmse_rf,
        rmse_gb
    ],
    "R2": [
        r2_linear,
        r2_rf,
        r2_gb
    ]
})
```

```python
results
```

```python
results.sort_values(
    by="R2",
    ascending=False
)
```

```python
best_model_name = results.loc[
    results["R2"].idxmax(),
    "Model"]
print("Best Model:", best_model_name)
```

```python
if best_model_name == "Linear Regression":
    best_model = linear_model
    best_predictions = y_pred_linear

elif best_model_name == "Random Forest":
    best_model = random_forest_model
    best_predictions = y_pred_rf

else:
    best_model = gradient_boosting_model
    best_predictions = y_pred_gb
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.5,color='red'
)

plt.xlabel("Actual Demand")
plt.ylabel("Predicted Demand")
plt.title("Actual vs Predicted Demand")

plt.show()
```

```python
prediction_comparison = pd.DataFrame({
    "Actual_Demand": y_test.values,
    "Predicted_Demand": best_predictions
})

prediction_comparison.head(20)
```

```python
errors = y_test.values - best_predictions
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    errors,
    bins=30
)

plt.xlabel("Prediction Error")
plt.ylabel("Frequency")
plt.title("Distribution of Prediction Errors")

plt.show()
```
